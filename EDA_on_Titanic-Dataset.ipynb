# Analysis of the Titanic dataset using Python and Pandas.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('https://raw.githubusercontent.com/itsluckysharma01/Datasets/refs/heads/main/Titanic_dataset.csv')
```

# Check For null Values

```python
df.isnull().sum()
```

# Handling Missing Values

```python
df.head()
```

Handling missing value in Age column by filling it with the median age.


```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df.isnull().sum()
```

Handling missing value in Fare column.


```python
df['Fare']=df['Fare'].fillna(df['Fare'].median())
```

```python
df.isnull().sum()
```

Handing missing value in Cabin column

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
df.isnull().sum()
```

Check for Duplicates values

```python
df.duplicated().sum()
```

# Bar Chart Survival rate by Class

```python
survival_by_Class= df.groupby('Pclass')['Survived'].mean()
survival_by_Class.plot(kind='bar', color=['blue', 'orange', 'green'])
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.xticks(rotation=0)
plt.show()
```

# Histogram Age Distribution

```python
sns.histplot(df['Age'], bins=30, kde=True, color='blue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

# Scatter plot of age vs fare

```python
plt.scatter(df['Age'], df['Fare'], alpha=0.5, color='green')
plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
